# 🤖 System A: Single-Agent Baseline Implementation
This notebook implements the baseline single-prompt LLM system to benchmark against the InsightForge multi-agent architecture.

```python
import time
import pandas as pd
import google.generativeai as genai

dbutils.widgets.text("gemini_key", "", "Gemini API Key")
api_key = dbutils.widgets.get("gemini_key")
genai.configure(api_key=api_key)

def run_single_agent_baseline(df: pd.DataFrame) -> dict:
    """
    System A: Single LLM call receiving raw dataset sample and asking for complete analysis.
    """
    start_time = time.time()
    
    # Construct raw sample context
    sample_text = df.head(10).to_string()
    info_text = f"Shape: {df.shape}\nColumns: {df.columns.tolist()}\nDtypes:\n{df.dtypes.to_string()}"
    
    prompt = f"""
    You are an expert data scientist and business strategist.
    Analyze the following dataset and provide:
    1. Schema & Data Overview
    2. Data Quality Issues (Nulls, Duplicates, Outliers)
    3. Exploratory Data Analysis & Statistics
    4. Business Insights & Recommendations
    
    Dataset Info:
    {info_text}
    
    Sample Data:
    {sample_text}
    """
    
    model = genai.GenerativeModel("gemini-1.5-flash")
    response = model.generate_content(prompt)
    elapsed_time = time.time() - start_time
    
    return {
        "system": "Single-Agent Baseline (System A)",
        "response": response.text,
        "elapsed_time_sec": round(elapsed_time, 2),
        "tokens_estimated": len(response.text.split())
    }
```
