# 🔬 InsightForge AI — Research Experiment Design
### Comparing Single-Agent LLM Prompting vs. Multi-Agent Orchestration for Automated Data Analysis

---

## 1. Research Question & Hypothesis
- **Research Question:** Does a multi-agent specialized architecture (LangGraph) generate higher-quality, deeper, and more reliable business insights than a single-turn monolithic LLM prompt?
- **Hypothesis (H1):** The multi-agent architecture will yield statistically significant improvements in **Insight Depth** and **Completeness** over single-agent baselines due to targeted contextual isolation and stage-wise refinement.
- **Hypothesis (H2):** The multi-agent architecture trades off higher latency (3-4x) for higher qualitative analysis scores.

---

## 2. Experimental Setup & Variables
- **Independent Variable:** Analysis Architecture
  - **System A:** Single-agent baseline (One LLM call with dataset summary & direct prompt)
  - **System B:** Multi-agent pipeline (InsightForge AI 7-agent LangGraph system)
- **Dependent Variables / Metrics:**
  1. **Insight Depth (1-5 scale):** Strategic richness, actionability, non-trivial deductions.
  2. **Completeness (1-5 scale):** Coverage of statistical distributions, correlations, outliers, and recommendations.
  3. **Response Time (seconds):** Total pipeline runtime from ingestion to final report.
  4. **Error / Hallucination Rate (%):** Frequency of incorrect statistical claims compared to ground truth.

---

4. **HR Employee Attrition Dataset** (1,470 rows × 10 cols) — Workforce churn, salary, performance ratings.
